## Decorator:

```python
import time
from functools import wraps


def process_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        elapsed_time = end_time - start_time
        print(f"Function '{func.__name__}' executed in {elapsed_time:.4f} seconds.")
        print("==================================================================")
        return result
    return wrapper


def log_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Executing function '{func.__name__}' with arguments: {args} and keyword arguments: {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function '{func.__name__}' returned: {result}")
        print("==================================================================")
        return result
    return wrapper

```

## Generator

```python

@process_time
@log_execution
def read_employee_data():
    yield from employee_data


@process_time
@log_execution
def filter_employees_by_department(department):
    for employee in read_employee_data():
        if employee["department"] == department:
            yield employee

@process_time
@log_execution
def filter_employees_by_age(min_age, max_age):
    for employee in read_employee_data():
        if min_age <= employee["age"] <= max_age:
            yield employee

@process_time
@log_execution
def highest_paid_employee():
    highest_paid = None
    for employee in read_employee_data():
        if highest_paid is None or employee["age"] > highest_paid["age"]:
            highest_paid = employee
    return highest_paid

```

```python
## Dataclass for Employee
from dataclasses import dataclass


@dataclass
class Employee:
    name: str
    age: int
    department: str

```

```python
## Read from employee_data and find the highest paid employee and filter employees by department and age using the Employee dataclass
emp_data = list(read_employee_data())
highest_paid = highest_paid_employee()
filtered_by_department = list(filter_employees_by_department("Engineering"))
filtered_by_age = list(filter_employees_by_age(25, 40))
print("=============== Highest Paid Employee and Filtered Employees ===============")
print("Highest Paid Employee:", highest_paid)
print("=====================Filtered Employees by Department=====================")
print("Filtered by Department (Engineering):", filtered_by_department)
print("=====================Filtered Employees by Age=====================")
print("Filtered by Age (25-40):", filtered_by_age)
```
